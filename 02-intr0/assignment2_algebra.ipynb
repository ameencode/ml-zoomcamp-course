A function that does vector-vector multiplication for two vectors u and v:

```python
def vector_vector_multiplication(u, v):
    assert u.shape[0] == v.shape[0], "Vectors must be of the same length for 1D array multiplication."

    n = u.shape[0]

    result = 0.0

    for val in range(n):
        result = result + u[val] * v[val]
    return result

vector_vector_multiplication(u, v)   
```

In NumPy, there is already a function that does this - dot:

```python
u.dot(v)
```

Matrix-vector multiplication

```python
U = np.array([
    [2, 4, 5, 6],
    [1, 2, 1, 2],
    [3, 1, 2, 1]
])
U.shape
```

Matrix_vector_multiplication

```python
def matrix_vector_multiplication(U, v):
    assert U.shape[1] == v.shape[0], "Matrix columns must match vector length for multiplication."

    m = U.shape[0]
    n = U.shape[1]

    result = np.zeros(m)

    for val in range(m):
        for j in range(n):
            result[val] += U[val, j] * v[j]
    return result

matrix_vector_multiplication(U, v)
```

Another method

```python
def matrix_vector_multiplication(U, v):
    assert U.shape[1] == v.shape[0]
    
    num_rows = U.shape[0]
    
    result = np.zeros(num_rows)
    
    for i in range(num_rows):
        result[i] = vector_vector_multiplication(U[i], v)
    
    return result

matrix_vector_multiplication(U, v)
```

Numpy Function

```python
U.dot(v)
```

Matrix_matrix multiplication

```python
V = np.array([
   [1, 1, 2],
   [0, 0.5, 1], 
   [0, 2, 1],
   [2, 1, 0],
 ])
```

```python
def matrix_matrix_multiplication(U, V):
    assert U.shape[1] == V.shape[0], "Number of columns in U must match number of rows in V for multiplication."

    m = U.shape[0]
    n = U.shape[1]
    p = V.shape[1]

    result = np.zeros((m, p))

    for i in range(m):
        for j in range(p):
            for k in range(n):
                result[i, j] += U[i, k] * V[k, j]
    
    return result

matrix_matrix_multiplication(U, V)
```

Another method

```python
def matrix_matrix_multiplication(U, V):
    assert U.shape[1] == V.shape[0]
    
    num_rows = U.shape[0]
    num_cols = V.shape[1]
    
    result = np.zeros((num_rows, num_cols))
    
    for i in range(num_cols):
        vi = V[:, i]
        Uvi = matrix_vector_multiplication(U, vi)
        result[:, i] = Uvi
    
    return result

matrix_matrix_multiplication(U, V)
```

Identity matrix
The identity matrix is a square matrix where on the diagonal we have ones, and zeros everywhere else:

Identity matrix is just like 1, we we multiply a matrix, say A, by an identity matrix, we get back the matrix A.

```python
# we use .eye() to create an identity matrix of size 3x3

I = np.eye(3)
I
```

```python
# to check if the identity matrix is working correctly, 
# we can multiply it with another matrix and see if we get the same matrix back.

V.dot(I)
```

Inverse
The identity matrix is useful for explaining what a matrix inverse is. Let's say we have a matrix A. The inverse of A, usually written A⁻¹, is a matrix such that when we multiply it by A, we get I.

First of all, only square matrices - matrices where the number of rows equals the number of columns - have an inverse.

```python
# V is a 4x3 matrix, so we need to create a 3x3 matrix from it. only square matrices can have an inverse.

Vs = V[[0, 1, 2]]
Vs
```

```python
# To get the inverse:

Vs_inv = np.linalg.inv(Vs)
Vs_inv
```

```python
# To verify that the inverse is correct, we can multiply Vs with its inverse:
# We should get the identity matrix back.

Vs.dot(Vs_inv)
```
